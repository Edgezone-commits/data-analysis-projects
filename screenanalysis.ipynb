# Sleep and Screen Time Analysis

**Tools:** Python, Pandas, NumPy, Matplotlib  
**Dataset:** [Kaggle - Sleep Mobile Stress](https://www.kaggle.com/datasets/jayjoshi37/sleep-screen-time-and-stress-analysis)
**Goal:** Analyze how phone usage and screen time affects 
sleep quality and stress levels

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('sleep_mobile_stress_dataset_15000.csv')
```

```python
## # Who uses phone the most before sleep?
```

```python
df.sort_values(by='phone_usage_before_sleep_minutes',ascending=False)
```

```python
df.sort_values(by='sleep_quality_score')
```

```python
## Most Stressed Peoples
```

```python
df.sort_values(by='stress_level',ascending=False).head(10)
```

```python
## Average stress per occupation
```

```python
df.groupby('occupation')['stress_level'].mean()
```

```python
## Average sleep hours per gender
```

```python
df.groupby('gender')['sleep_duration_hours'].mean()
```

```python
## Average screen time per occupation
```

```python
df.groupby('occupation')['daily_screen_time_hours'].mean()
```

```python
# Max mental fatigue per occupation
```

```python
df.groupby('occupation')['mental_fatigue_score'].max()
```

```python
# Count of people per occupation
```

```python
df.groupby('occupation')['user_id'].count()
```

```python
# Compare sleep quality and stress across occupations
```

```python
pd.pivot_table(df,values=['sleep_quality_score','stress_level','mental_fatigue_score'],index='occupation',aggfunc='mean')
```

```python
# Compare screen time and sleep by gender
```

```python
pd.pivot_table(df,values=['daily_screen_time_hours','sleep_duration_hours'],index='gender',aggfunc='mean')
```

```python
# Label stress as High or Low
```

```python
df['stress_level'].apply(lambda x: 'High' if x > 7 else 'low')
```

```python
## Label sleep as Good or Bad
```

```python
df['sleep_quality_score'].apply(lambda x: 'Good' if x >= 7 else 'Bad')
```

```python
# Convert screen time hours to minutes
```

```python
df['daily_screen_time_hours'].apply(lambda x: x *60)
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
df.to_csv('updated_screentime_dataset.csv',index = False)
```

```python
import matplotlib.pyplot as plt
```

```python
# 1. Screen Time vs Sleep Quality
```

```python
plt.scatter(df['daily_screen_time_hours'], df['sleep_quality_score'])
plt.title('Screen Time vs Sleep Quality')
plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Sleep Quality Score')
plt.show()
```

```python
# 1. Distribution of Sleep Hours
```

```python
df['sleep_duration_hours'].plot(kind='hist')
plt.title('Distribution of Sleep Hours')
plt.xlabel('Sleep Duration (Hours)')
plt.ylabel('Count')
plt.show()
```

```python
df['mental_fatigue_score'].plot(kind='hist')
plt.title('Distribution of Mental Fatigue')
plt.xlabel('Mental Fatigue Score')
plt.ylabel('Count')
plt.show()
```

```python
df['sleep_duration_hours'].plot(kind='hist', bins=20, color='green', edgecolor='black')
plt.show()
```

```python
# 1. Average Screen Time per Occupation
```

```python
df.groupby('occupation')['daily_screen_time_hours'].mean().plot(kind='barh',color='orange')
plt.title('Average Screen Time per Occupation')
plt.xlabel('Occupation')
plt.ylabel('Average Screen Time(Hours)')
plt.show()
```

```python
df.groupby('occupation')['phone_usage_before_sleep_minutes'].mean().plot(kind='barh',color='blue')
plt.title('occupation and phone usages')
plt.xlabel('Occupation')
plt.ylabel('Phone Usage')
plt.show()
```
