```python
## # Who uses phone the most before sleep?
```

```python
df.sort_values(by='phone_usage_before_sleep_minutes',ascending=False)
```

```python
## Who has the worst sleep quality?
```

```python
df.sort_values(by='sleep_quality_score')
```

```python
## Most Stressed Peoples
```

```python
df.sort_values(by='stress_level',ascending=False).head(10)
```

```python
## Average stress per occupation
```

```python
df.groupby('occupation')['stress_level'].mean()
```

```python
## Average sleep hours per gender
```

```python
df.groupby('gender')['sleep_duration_hours'].mean()
```

```python
## Average screen time per occupation
```

```python
df.groupby('occupation')['daily_screen_time_hours'].mean()
```

```python
# Max mental fatigue per occupation
```

```python
df.groupby('occupation')['mental_fatigue_score'].max()
```

```python
# Count of people per occupation
```

```python
df.groupby('occupation')['user_id'].count()
```

```python
# Compare sleep quality and stress across occupations
```

```python
pd.pivot_table(df,values=['sleep_quality_score','stress_level','mental_fatigue_score'],index='occupation',aggfunc='mean')
```

```python
# Compare screen time and sleep by gender
```

```python
pd.pivot_table(df,values=['daily_screen_time_hours','sleep_duration_hours'],index='gender',aggfunc='mean')
```

```python
# Label stress as High or Low
```

```python
df['stress_level'].apply(lambda x: 'High' if x > 7 else 'low')
```

```python
## Label sleep as Good or Bad
```

```python
df['sleep_quality_score'].apply(lambda x: 'Good' if x >= 7 else 'Bad')
```

```python
# Convert screen time hours to minutes
```

```python
df['daily_screen_time_hours'].apply(lambda x: x *60)
```
