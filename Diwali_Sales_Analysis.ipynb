## 7. Data Cleaning

Cleaning the dataset by removing irrelevant columns, handling 
missing values and creating new useful columns for analysis.

### 7.1 Dropping Irrelevant Columns

Dropping Cust_name, User_ID and unnamed1 as they do not 
contribute to the analysis. Status column is also entirely 
empty so it will be dropped too.

```python
# Dropping columns that are irrelevant or entirely empty
df = df.drop(columns=['Cust_name', 'User_ID', 'unnamed1', 'Status'])

# Checking it
df.columns
```

### 7.2 Handling Missing Values

Removing rows where Amount is missing since it is a 
critical column for all sales analysis.

```python
# Checking missing values before cleaning
df.isna().sum()
```

```python
# Remove rows with missing Amount values
df = df.dropna(subset=['Amount'])

# Confirm no missing values remain
df.isnull().sum()
```

### 7.3 Converting Age Group to Ordinal Categories

Converting Age Group from raw string ranges to proper 
ordered categories so it can be used correctly in analysis.

```python
# Age group ranges to meaningful category names
age_group = {
    '0-17': 'Teen',
    '18-25': 'Young Adult',
    '26-35': 'Adult',
    '36-45': 'Middle Aged',
    '46-50': 'Senior',
    '51-55': 'Senior',
    '55+': 'Senior'
}

# Apply the mapping to create proper categories
df['Age Group'] = df['Age Group'].map(age_group)

# Convert to ordered categorical variable
df['Age Group'] = pd.Categorical(df['Age Group'],
                  categories=['Teen', 'Young Adult', 'Adult', 'Middle Aged', 'Senior'],
                  ordered=True)

# Checking the conversion
print(df['Age Group'].value_counts())
```
